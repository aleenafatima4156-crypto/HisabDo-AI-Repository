```python
import pandas as pd
students={
    "Name":[
        "Ali", "ALia","Ahmed","Ayesha", "Fatima","Abdullah","Sara","Hania","Hadia","Zain"],
    "Age": [21,18,18,16,17,22,24,20,18,20],
    "Course":["AI","ML","AI","DSA","COAL","ML","Cyber security","DSA","Data science","COAL"],
    "Marks":[85,75,60,70,89,90,98,89,45,50]
}
```

```python
dataset =pd.DataFrame(students)
```

```python
print ("All students")
print (dataset)
```

```python
print("students with marks above 70")
print(dataset[dataset["Marks"]>70])
```

```python
avg_marks=dataset["Marks"].mean()
print("average marks: ",avg_marks)
```

```python
highest= dataset.loc[dataset["Marks"].idxmax()]
print ("highest marks:")
print(highest)
```

```python
lowest= dataset.loc[dataset["Marks"].idxmin()]
print ("lowest marks:")
print(lowest)
```

```python
total_students=len(dataset)
print("total students: ",total_students)
```
