# import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

# Load Dataset

```python
df = pd.read_csv(r"C:\Users\human\Downloads\student_performance.csv")
```

# Basic Information

```python
# Basic Information
print("information of dataset")
print(df.info())
```

```python
print("first 5 rows")
print(df.head())
```

```python
print("statistical information of dataset")
print(df.describe())
```

```python
print("unique values in data")
print(df.nunique())
```

# find Missing Values

```python
df.isnull().sum()
```

# Handle Missing Values

```python
df.fillna(df.select_dtypes(include='number').mean(), inplace=True)
df
```

# Average Scores

```python
print("\nAverage Assignment Score:", df["Assignment_Score"].mean())
print("Average Midterm Score:", df["Midterm_Score"].mean())
print("Average Final Score:", df["Final_Score"].mean())
```

# Highest and Lowest Final Score

```python
highest = df.loc[df["Final_Score"].idxmax()]
lowest = df.loc[df["Final_Score"].idxmin()]

print("\nHighest Score Student")
print(highest)

print("\nLowest Score Student")
print(lowest)
```

# Attendance below 75%

```python
low_attendance = df[df["Attendance"] < 75]
print("\nStudents with Attendance below 75%")
print(low_attendance)
```

# At Risk Students

```python
risk_students = df[df["Final_Score"] < 60]
print("\nStudents at Risk of Failing")
print(risk_students)
```

# Average Score by Course

```python
course_avg = df.groupby("Course")["Final_Score"].mean()
print("\nAverage Final Score by Course")
print(course_avg)
```

# Relationship between Attendance and Final Score

```python
correlation = df["Attendance"].corr(df["Final_Score"])
print("\nCorrelation:", correlation)
```

# Visualization 1 
histogram graph

score distribution chart

```python
plt.hist(df["Final_Score"], bins=10)
plt.title("Score Distribution")
plt.xlabel("Final Score")
plt.ylabel("Students")
plt.show()

```

# Visualization 2
bar graph

avg score by course

```python
course_avg.plot(kind="bar")
plt.title("Average Score by Course")
plt.ylabel("Average Final Score")
plt.show()

```

# Visualization 3
scatter graph

```python
plt.scatter(df["Attendance"], df["Final_Score"])
plt.title("Attendance vs Final Score")
plt.xlabel("Attendance")
plt.ylabel("Final Score")
plt.show()
```
