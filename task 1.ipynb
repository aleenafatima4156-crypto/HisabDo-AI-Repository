```python
def display_student(name, age, skills):
    print ("\n--- student information---")
    print ("Name: ", name)
    print ("Age: ", age)
    print ("skills: ", skills)
```

```python
name= input("enter your name:")
age = int(input("enyer your age:"))
skills=[]
num_skills=int(input("how many skills do you want to add?"))

for i in range(num_skills):
    skill= input(f"enter skills {i+1}:")
    skills.append(skill)

if age>=18:
    status="adult"
else:
    status="minor"

```

```python
display_student(name, age, skills)
print("status:", status)
print("\nskills list:")
for skill in skills:
    print("-", skill)

print ("\nthank you for using the student information system!")
```
